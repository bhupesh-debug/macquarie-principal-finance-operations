# 01 — Exploratory Data Analysis

Synthetic loan-portfolio operations data: 3 product lines x 730 days, a 600-facility loan book, and a daily pipeline-health file with 25 injected data-quality incidents.

```python
import sys
sys.path.insert(0, '../src')
import pandas as pd
ops, health, fac = __import__('data_pipeline').load()
print(ops.shape, health.shape, fac.shape)
print(ops.groupby('product_line')['deals_received'].describe())
```

```python
import matplotlib.pyplot as plt
tot = __import__('data_pipeline').daily_totals(ops)
tot.plot(x='date', y=['deals_received', 'deals_closed'], figsize=(10, 4), title='Daily pipeline volume')
plt.show()
```

**What to look for:** the weekly cycle (weekend troughs), the slow growth trend, and the gap between received and closed — that gap is the backlog this project manages.
