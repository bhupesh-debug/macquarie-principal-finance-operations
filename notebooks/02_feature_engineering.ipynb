# 02 — Feature Engineering

Forecasting features use only information available at forecast time. Same-day operational features are EXCLUDED from the forecaster (leakage) — they belong to the driver model.

```python
import sys
sys.path.insert(0, '../src')
import data_pipeline as dp
import feature_engineering as fe
ops, _, _ = dp.load()
tot = dp.daily_totals(ops)
feat = fe.add_forecast_features(tot, 'deals_closed')
print(feat[fe.FORECAST_FEATURES].head())
train, test, cut = fe.time_split(
    feat.dropna().reset_index(drop=True), 61)
print('hold-out starts:', cut.date())
```

**Leakage check:** no same-day `staffing_ratio`, `backlog_ratio` or `exception_rate_pct` in `FORECAST_FEATURES`. If one appears, the forecast is invalid — fix it before modelling.
