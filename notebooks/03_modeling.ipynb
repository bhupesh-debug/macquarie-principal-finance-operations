# 03 — Modeling

Five forecasters on the same features and the same 61-day hold-out: seasonal naive, Holt-Winters (additive, hand-rolled), linear regression, random forest, gradient boosting.

```python
import sys
sys.path.insert(0, '../src')
import data_pipeline as dp
import feature_engineering as fe
import models as md
ops, _, _ = dp.load()
tot = dp.daily_totals(ops)
feat = fe.add_forecast_features(tot, 'deals_closed')
train, test, cut = fe.time_split(
    feat.dropna().reset_index(drop=True), 61)
fitted = md.fit_sklearn_models(
    train[fe.FORECAST_FEATURES], train['deals_closed'])
for name, m in fitted.items():
    print(name, md.metrics(test['deals_closed'],
          m.predict(test[fe.FORECAST_FEATURES])))
```

**Expected result:** Random Forest wins on deals closed (MAPE ~22.6%); Linear wins on cycle time (MAPE ~4.1%). Holt-Winters trails — MAPE is inflated by weekend troughs where closings fall to single digits.
